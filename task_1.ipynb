# Task 6 — Final Insights & Business Recommendations

## Key Insights

1. Customer churn is an important business issue, with a significant portion of customers leaving the company.

2. Contract type shows differences in churn behavior. Customers with shorter-term contracts generally require more attention for retention.

3. Internet service type is associated with different churn levels, making service quality and customer experience important retention factors.

4. Payment method shows differences in churn rates. Customers using higher-churn payment methods can be targeted for retention campaigns.

5. Tenure is strongly related to churn behavior. Newer customers require special attention during their early period with the company.

6. Monthly charges differ between customers who churn and customers who stay, indicating that pricing and perceived value may influence customer decisions.

## Business Recommendations

- Offer incentives to customers on short-term contracts to encourage longer contracts.
- Provide special retention offers to new customers during their first year.
- Improve customer support and technical assistance for customers experiencing service problems.
- Identify high-churn customer segments and provide personalized retention offers.
- Review pricing and service packages for customers with higher monthly charges.
- Encourage customers to adopt services that improve their overall customer experience.
- Regularly monitor churn rates using dashboards and customer analytics.

## Final Conclusion

The analysis shows that customer churn is influenced by several customer, service, contract, and billing characteristics.

Data analysis can help the company identify high-risk customer segments and take targeted actions to improve customer retention and reduce churn.
